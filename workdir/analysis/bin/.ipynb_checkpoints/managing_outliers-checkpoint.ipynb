```python
import sklearn
import numpy as np
import scipy as sp
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# pd.set_option('display.max_colwidth', None)
# pd.set_option('display.width', 1000)
# pd.set_option('display.expand_frame_repr', False)  # Disable line-wrapping
# pd.set_option('display.max_rows', None)


coffea_caff = pd.read_csv(r"..\data\coll_caff_node_w_class.csv")
coffee_data = coffea_caff.copy()

# coffea_caff_clim = pd.read_csv(r"..\data\coll_caff_node_clim_w_class.csv")
# coffee_data = coffea_caff_clim.copy()

#coffea_caff_env = pd.read_csv(r"..\data\coll_caff_node_env_w_class.csv")
#coffee_data = coffea_caff_env.copy()

#distinct_vals = coffee_data['caffeine_class'].value_counts()
#print(distinct_vals)

coffee_data.drop(columns=['specimen_id','longitude','latitude', 'source_crs', 'mada_geom_point', 'sampled_layers', 'nodata_layers', 'is_categorical_encoded'],inplace=True)

coffee_data.head
```

Renaming headers

```python

coffee_data.columns = (coffee_data.columns
              .str.replace(r'[^\w\s]', '', regex=True)  # Remove special characters like °, parentheses, hyphen
              .str.replace(r'\s+', '_', regex=True)     # Replace spaces with underscores
              .str.strip())                             # Strip leading and trailing spaces


mapping_df = pd.read_csv(r'../data/column_mapping.csv')

mapping_df['old_name'] = (mapping_df['old_name']
                          .str.replace(r'[^\w\s]', '', regex=True)  # Clean the old names similarly
                          .str.replace(r'\s+', '_', regex=True)     # Replace spaces with underscores
                          .str.strip())
mapping_df

# Create the mapping dictionary
mapping_dict = dict(zip(mapping_df['old_name'], mapping_df['new_name']))

# Rename columns in the DataFrame using the cleaned mapping
coffee_data.rename(columns=mapping_dict, inplace=True)
coffee_data.columns
```

```python
clim_columns = list(coffee_data.filter(like='clim_').columns) + [ 'env_71_alt', 'env_72_slo', 'env_73_asp', 'env_74_solrad', 'env_79_forcov' ]

clim_data = coffee_data[clim_columns]

env_data = coffee_data.drop(columns=clim_columns+['caffeine_percent', 'caffeine_class'])

# Optional: View the column names for confirmation
print("Clim Data Numerical Features:")
print(clim_data.columns)

print("\nEnv Data Categorical Features:")
print(env_data.columns)
```

```python
missing_data = coffee_data.isnull().sum()
print(missing_data)
```

**Amongst the numerical values**, we Removed the 11 clim_ columns with NaNs and Inserted median value in place of NaN for env_columns

```python
# Identify columns that start with 'clim_' and 'env_'
clim_columns = [col for col in clim_data.columns if col.startswith('clim_')]
env_columns = [col for col in clim_data.columns if col.startswith('env_')]

# Remove rows where any 'clim_' column has NaN
coffee_data = coffee_data.dropna(subset=clim_columns)

# Fill missing values in 'env_' columns with the median
for col in env_columns:
    median_value = coffee_data[col].median()
    coffee_data[col] = coffee_data[col].fillna(median_value)

# Display the updated dataframe
#coffee_data.shape
missing_data = coffee_data.isnull().sum()
missing_data
```

Exploring and Managing outliers


```python

def plot_violin_with_boxplots(data, cols):
    # Ensure we are working with only numeric columns
    numeric_cols = data[cols].select_dtypes(include='number').columns

    num_plots = len(numeric_cols)
    num_rows = (num_plots + 1) // 2  # Number of rows (2 plots per row)
    num_cols = 4  # 2 columns for side-by-side plots

    # Set up the grid of subplots
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, 4 * num_rows))
    axes = axes.flatten()  # Flatten the axes array for easier indexing

    for i, column in enumerate(numeric_cols):
        # Remove NaN values from the column to avoid plotting errors
        col_data = data[column].dropna()

        # Plot the violin plot with the boxplot integrated
        sns.violinplot(y=col_data, inner="box", ax=axes[i])
        axes[i].set_title(f'Distr of {column}')
        axes[i].set_xlabel('')  # Optional: clear x-label for clarity
        axes[i].set_ylabel(column)

    # Hide any unused axes
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])

    # Adjust layout for better spacing
    plt.tight_layout()
    plt.show()
```

```python
plot_violin_with_boxplots(clim_data, clim_data.columns)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

def plot_distribution_and_boxplots(data, cols):
    num_plots = len(cols)
    num_rows = (num_plots + 1) // 2  # Determine the number of rows
    num_cols = 2  # We will plot two columns side by side (distribution and boxplot)

    # Set up the grid of subplots
    fig, axes = plt.subplots(num_rows * 2, num_cols, figsize=(15, 4 * num_rows))  # Allocate 2 rows per feature
    
    axes = axes.flatten()  # Flatten the axes array to easily access each subplot
    
    for i, column in enumerate(cols):
        # Get the indices for the plots (each column has 2 subplots: distribution and boxplot)
        dist_plot_idx = 2 * i  # Index for distribution plot
        box_plot_idx = dist_plot_idx + 1  # Index for boxplot
        
        # Plot the distribution (histogram and KDE)
        sns.histplot(data[column], kde=True, ax=axes[dist_plot_idx])
        axes[dist_plot_idx].set_title(f'Distribution of {column}')
        axes[dist_plot_idx].set_ylabel('Frequency')
        axes[dist_plot_idx].set_xlabel(column)
        
        # Plot the boxplot
        sns.boxplot(data=data, x=column, ax=axes[box_plot_idx])
        axes[box_plot_idx].set_title(f'Boxplot of {column}')
        axes[box_plot_idx].set_xlabel(column)
        axes[box_plot_idx].set_ylabel('')

    # Adjust layout for better spacing
    plt.tight_layout()
    #plt.show()
    plt.savefig(r"../images/outliers_histo.png")
```

```python
plot_distribution_and_boxplots(clim_data, clim_data.columns)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Assuming clim_data is a DataFrame with all numerical columns
num_cols = clim_data.shape[1]  # Number of numerical columns
cols = 3  # Number of columns for subplots grid
rows = int(np.ceil(num_cols / cols))  # Determine the number of rows based on the number of features

fig, axes = plt.subplots(rows, cols, figsize=(15, rows * 5))
axes = axes.flatten()  # Flatten the 2D array of axes for easy iteration

for i, column in enumerate(clim_data.columns):
    axes[i].boxplot(clim_data[column])
    axes[i].set_title(f'Boxplot of {column}')
    axes[i].tick_params(axis='x', which='both', bottom=False, top=False, labelbottom=False)

# Remove unused axes if any
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define the number of rows and columns for subplots
num_cols = clim_data.shape[1]  # Number of features
cols = 3  # Number of columns in the grid
rows = int(np.ceil(num_cols / cols))  # Number of rows based on number of columns

fig, axes = plt.subplots(rows, cols, figsize=(15, rows * 5))
axes = axes.flatten()  # Flatten axes for easy iteration

for i, column in enumerate(clim_data.columns):
    # Create violin plot (in light blue)
    sns.violinplot(x=clim_data[column], ax=axes[i], inner=None, color='#ADD8E6')  # Light blue (Hex code for lightblue)
    
    # Overlay the boxplot (in dark blue)
    sns.boxplot(x=clim_data[column], ax=axes[i], color='#1E90FF', width=0.2, 
                boxprops=dict(alpha=0.6, zorder=3), whiskerprops=dict(zorder=3),
                capprops=dict(zorder=3), medianprops=dict(zorder=3), flierprops=dict(zorder=3))

    axes[i].set_title(f'{column}')

plt.tight_layout()
#plt.show()
plt.savefig(r"../images/outliers_box_violon.png")

```
