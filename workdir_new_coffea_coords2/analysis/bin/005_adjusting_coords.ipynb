```python
import pandas as pd

# Load the dataset
file_path = r"..\..\..\workdir\input\0026013-240906103802322.csv"
df = pd.read_csv(file_path, sep='\t', engine='python')

df_mg = df[df['countryCode'] == 'MG']
```

```python
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', 1000)
pd.set_option('display.expand_frame_repr', False)  # Disable line-wrapping
pd.set_option('display.max_rows', None)
```

```python
# Define ex situ clues
ex_situ_keywords = ['botanic', 'garden', 'cultivated', 'nursery', 'ex situ', 'greenhouse', 'collection']
ex_situ_institutions = ['RBGK', 'MO', 'PBG', 'BR', 'TEF', 'TAN'] 

# Build masks

mask_locality = df_mg['locality'].astype(str).str.contains('|'.join(ex_situ_keywords), case=False, na=False)
mask_institution = df_mg['institutionCode'].astype(str).isin(ex_situ_institutions)
mask_basisOfRecord = df_mg['basisOfRecord'].astype(str).isin(['LIVING_SPECIMEN'])  # optional stricter filter

# Combine masks: remove records where any mask is True
mask_ex_situ = mask_locality | mask_institution | mask_basisOfRecord

# Filter
df_filtered = df_mg[~mask_ex_situ]

# Save to new file
output_path = r'..\data\0026013-240906103802322_filtered.csv'
df_filtered.to_csv(output_path, index=False)

# Summary
print(f"Original records: {len(df)}")
print(f"MG records: {len(df_mg)}")
print(f"Filtered records: {len(df_filtered)}")
print(f"Filtered data saved to: {output_path}")
```

```python
counts_df = (
    df_mg
    .groupby('species')   # ou le nom exact de ta colonne espèce
    .size()               # compte le nombre de lignes par groupe
    .reset_index(name='count')
    .sort_values(by='count', ascending=False)
)

counts_df
```

```python
counts = df_mg['species'].value_counts()
total_from_counts = counts.sum()

counts_df = (
    df_mg
    .groupby('species')
    .size()
    .reset_index(name='count')
)
total_from_counts_df = counts_df['count'].sum()

# In either case, compare to the number of rows:
n_rows = len(df_mg)

print(f"Sum of counts (value_counts): {total_from_counts}")
print(f"Sum of counts (groupby):      {total_from_counts_df}")
print(f"Number of rows in df_mg:      {n_rows}")

```

```python
n_missing = df_mg['species'].isna().sum()

# 2. Nombre de chaînes vides ou espaces seulement
n_empty_str = df_mg['species'].fillna('').astype(str).str.strip().eq('').sum()

print(f"NaN in 'species' column:            {n_missing}")
print(f"Empty strings ('' or spaces only): {n_empty_str}")
print(f"Total rows in df_mg:               {len(df_mg)}")
```
